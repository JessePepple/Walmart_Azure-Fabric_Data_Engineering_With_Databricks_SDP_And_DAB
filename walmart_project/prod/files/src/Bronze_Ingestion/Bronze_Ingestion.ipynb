```python
dataset = ["Products", "Customers", "Employees", "Orders", "Order_Items", "Stores"]

for data in dataset:
    df = spark.readStream.format("cloudFiles").option("cloudFiles.format", "parquet").option("cloudFiles.inferColumnTypes", "true").option("cloudFiles.schemaLocation", f"abfss://silver@walmartlakejess.dfs.core.windows.net/{data}_Enr/Checkpoint_Location/schema/").option("cloudFiles.schemaEvolutionMode", "addNewColumns").load(f"abfss://bronze@walmartlakejess.dfs.core.windows.net/{data}/")

    df.writeStream.format("delta").outputMode("append").option("checkpointLocation", f"abfss://silver@walmartlakejess.dfs.core.windows.net/{data}_Enr/Checkpoint_Location/checkpoint").trigger(once=True).toTable(f"walmart_catalog.bronze.{data}")
```
